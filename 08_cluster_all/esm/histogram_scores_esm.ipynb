```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
file = "/workspaces/Effector-Structure-Prediction/08_cluster_all/esm/effector_p_esm_foldseek_results.tsv"
df = pd.read_csv(file, sep="\t", header=None)

similarities = df.iloc[:, 4]

plt.hist(similarities, bins=50)
plt.xlabel("Query tmscore")
plt.ylabel("Number of comparisons")
plt.title("Distribution of Effector P ESM Foldseek similarities")
plt.show()
```
